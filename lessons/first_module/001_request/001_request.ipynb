## 3. Create the client and choose a model

`Anthropic()` automatically picks up `ANTHROPIC_API_KEY` from the environment, so you don't pass the key in.

Model options (latest family):
| Model | ID | Use when |
|---|---|---|
| Opus | `claude-opus-5-5` | Hardest tasks, best quality |
| Sonnet | `claude-sonnet-5-5` | Great balance of quality, speed and cost |
| Haiku | `claude-haiku-5-5` | Fastest and cheapest |

````python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5-5"
````

## 4. Send a request

`client.messages.create()` takes three required arguments:
- **`model`** – which Claude model to use
- **`max_tokens`** – the maximum length of the reply. This is a safety cap, not a target: Claude stops when it's done or when it hits this limit.
- **`messages`** – the conversation so far, as a list of `{"role": ..., "content": ...}` dicts. `role` is `"user"` (you) or `"assistant"` (Claude).

````python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence.",
        }
    ],
)

message
````

## 5. Read the response

The full `message` object above has metadata (`id`, `model`, `stop_reason`, `usage`, ...).
The text itself is in `message.content`, which is a **list of content blocks**. For a plain text reply, the first block holds the answer:

````python
message.content[0].text
````

## 6. Useful metadata

- `stop_reason` – why Claude stopped: `"end_turn"` (finished naturally) or `"max_tokens"` (hit your cap, so the answer may be cut off)
- `usage` – input/output token counts, which is what you're billed for

````python
print("Stop reason:  ", message.stop_reason)
print("Input tokens: ", message.usage.input_tokens)
print("Output tokens:", message.usage.output_tokens)
````
